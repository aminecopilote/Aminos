```python
import torch

print("GPU disponible :", torch.cuda.is_available())

if torch.cuda.is_available():
    print("Modèle :", torch.cuda.get_device_name(0))
    total = torch.cuda.get_device_properties(0).total_memory
    print(f"VRAM : {total / 1024**3:.1f} Gio")
else:
    print("Vérifiez que le type d’exécution utilise un GPU.")
```

```python
import time
import torch

if torch.cuda.is_available():
    a = torch.randn((2048, 2048), device="cuda")
    torch.cuda.synchronize()
    debut = time.perf_counter()

    resultat = a @ a
    torch.cuda.synchronize()

    print(f"Calcul GPU réussi en {time.perf_counter() - debut:.3f} seconde(s)")

    del a, resultat
    torch.cuda.empty_cache()
else:
    print("Aucun GPU disponible pour ce test.")
```
